```python
import matplotlib.pyplot as plt
```

```python
#data

x=[1,2,3,4,5]
y=[33,44,50,66,90]
```

```python
plt.plot(x,y)
plt.show()
```

```python
from matplotlib.lines import lineStyles
plt.figure(figsize=(6,4))
plt.plot(x,y,color='b',linestyle='--', marker='o',markersize='10')
plt.title('PLOT')
plt.xlabel('x-axis')
plt.ylabel('y-axis')
plt.show()
```

```python
x=[1,2,3,4,5]
y1=[33,44,50,66,90]
y2=[33,44,50,66,90]

plt.plot(x,y1,label='Sales 2024')
plt.plot(x,y2,label='Sales 2025')
plt.title('PLOT')
plt.xlabel('x-axis')
plt.ylabel('y-axis')
plt.legend()
plt.show()
```

```python
x=['A', 'B', 'C', 'D', 'E']
y=[33,44,95,66,90]
plt.bar(x,y)
plt.title('Bar chart')
plt.show()
```

```python
import random
data= [random.randint(1,20) for _ in range(100)]
plt.hist(data, bins=20)
plt.title('Histogram')
plt.show()
```

```python
x=['A', 'B', 'C', 'D', 'E']
y=[33,44,95,66,90]
plt.pie(y, labels=x, autopct='%1.1f%%', startangle=90)
plt.title('Pie chart')
plt.show()
```

```python
x=[10,24,36,48,53]
y=[33,44,50,66,90]

plt.scatter(x,y)
plt.title('Scatter plot')
plt.show()
```
