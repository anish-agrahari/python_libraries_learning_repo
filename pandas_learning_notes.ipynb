**What this cell does:** Selects the first row using `iloc[0]`, then accesses the `Country` value using the column name.

```python
df[df["Category"]=="Electronics"]
```

## 18. Filter rows by one condition

**What this cell does:** Keeps only rows where the `Category` column is equal to `Electronics`.

Pattern:
`df[condition]`

This is called **Boolean filtering**.

```python
df[(df["Category"]=="Electronics") & (df["Country"] == "USA")]
```

## 19. Filter using AND (`&`)

**What this cell does:** Keeps rows where **both** conditions are true:

1. `Category` is `Electronics`
2. `Country` is `USA`

`&` means **AND** in Pandas Boolean filtering.

```python
df[(df["Category"]=="Electronics") | (df["Country"] == "USA")]
```

## 20. Filter using OR (`|`)

**What this cell does:** Keeps rows where **at least one** condition is true:

1. `Category` is `Electronics`, OR
2. `Country` is `USA`

`|` means **OR** in Pandas Boolean filtering.

```python
df[df['Price'] > 200]
```

## 21. Filter using a numeric condition

**What this cell does:** Selects rows where `Price` is greater than `200`.

```python
df[df['CustomerName'].str.startswith("A")]
```

## 22. Filter strings that start with a value

**What this cell does:** `.str.startswith("A")` checks whether each `CustomerName` starts with the letter `A`.

The rows satisfying the condition are returned.

```python
df[df['CustomerName'].str.endswith('n')]
```

## 23. Filter strings that end with a value

**What this cell does:** `.str.endswith('n')` checks whether each `CustomerName` ends with the letter `n`.

```python
df[df['Country'].isin(["USA", "UK", "Germany", "Sweden"])]
```

## 24. Filter using isin()

**What this cell does:** `isin([...])` checks whether each country belongs to the given list of countries.

Here it keeps rows whose country is one of: `USA`, `UK`, `Germany`, or `Sweden`.

```python
df[~df['Country'].isin(["USA", "UK", "Germany", "Sweden"])]
```

## 25. Reverse an isin() condition

**What this cell does:** `~` reverses the Boolean condition.

So instead of selecting countries in the list, this selects countries **not** in the list.

```python
df.loc[df['Country'] == "USA"]
```

## 26. Filter rows using loc

**What this cell does:** `.loc[...]` is label-based selection. Here it selects all rows where `Country` is `USA`.

```python
df.loc[df['Country'] == "USA", 'Product']
```

## 27. Select a column after filtering with loc

**What this cell does:** First filters rows where `Country` is `USA`, then selects only the `Product` column.

So the result contains the products ordered from USA.

```python
df.loc[df['Country'] == "USA", 'Product']
```

## 28. Same operation again

**What this cell does:** This cell contains the same code as the previous cell. It again selects the `Product` values for rows where `Country` is `USA`.

```python
df.loc[df['Country'] == "USA", 'Product'] = 'Laptop'
```

## 29. Update values using loc

**What this cell does:** Finds rows where `Country` is `USA` and changes their `Product` value to `Laptop`.

`.loc[condition, column] = value` can be used to update selected DataFrame values.

```python
df.loc[df['Country'] == "USA", 'Product']
```

**What this cell does:** Runs the same USA + Product selection again so you can see the values after the update made in the previous cell.
