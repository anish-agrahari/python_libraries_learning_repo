# Seaborn Visualization Cheatsheet

This notebook serves as a revision guide for common Seaborn visualizations using the `penguins` dataset. Each section will explain the plot type, its purpose, and key parameters.

```python
import seaborn as sns
import pandas as pd
import numpy as np
```

## 1. Setup and Data Loading

First, we import the necessary libraries: `seaborn` for plotting, `pandas` for data manipulation, and `numpy` for numerical operations.

```python
sns.get_dataset_names()
```

### Listing Available Datasets

`sns.get_dataset_names()` lists all built-in datasets available in Seaborn, which can be useful for examples and learning.

```python
penguins = sns.load_dataset("penguins")
```

### Loading the Penguins Dataset

`sns.load_dataset('penguins')` loads the 'penguins' dataset, a popular dataset for demonstrating data visualization. It's automatically loaded into a pandas DataFrame.

```python
penguins.head()
```

## 2. Data Inspection

We start by inspecting the DataFrame to understand its structure and content.

```python
penguins.tail()
```

### Displaying Tail of the DataFrame

`DataFrame.tail()` shows the last 5 rows of the DataFrame, which is useful for checking the end of the dataset, especially after operations that might affect the last entries.

```python
penguins
```

### Displaying the Full DataFrame

Simply calling the DataFrame name (`penguins`) displays its entire content, or a truncated version if it's too large for display.

```python
penguins['species'].value_counts()
```

### Value Counts for Categorical Columns

`DataFrame['column'].value_counts()` returns a Series containing counts of unique values in the specified column. This is crucial for understanding the distribution of categorical features.

```python
penguins['island'].value_counts()
```

## 3. Scatter Plots

Scatter plots are used to visualize the relationship between two numerical variables. Each point represents an observation.

```python
sns.scatterplot(data=penguins, x='flipper_length_mm', y='body_mass_g')
```

### Basic Scatter Plot

`sns.scatterplot()`:
- `data`: The DataFrame containing the data.
- `x`: Column name for the x-axis.
- `y`: Column name for the y-axis.

```python
sns.scatterplot(data=penguins, x='flipper_length_mm', y='body_mass_g', hue='species')
```

### Scatter Plot with `hue`

- `hue`: Column name to color points based on categories in this column. Useful for visualizing relationships across different groups.

Here, `hue='species'` colors the points by penguin species, revealing distinct clusters.

```python
sns.scatterplot(data=penguins, x='flipper_length_mm', y='body_mass_g', hue='sex')
```

### Scatter Plot with `hue` (Sex)

Using `hue='sex'` to differentiate between male and female penguins.

```python
sns.set_style('ticks')
sns.scatterplot(data=penguins, x='flipper_length_mm', y='body_mass_g', hue='island')
```

## 4. Customizing Plot Aesthetics

Seaborn allows extensive customization of plot styles and contexts to enhance readability and presentation.

```python
sns.set_style('darkgrid') #dark #darkgrid
sns.scatterplot(data=penguins, x='flipper_length_mm', y='body_mass_g', hue='island')
```

### Setting Plot Style with `sns.set_style()`

`sns.set_style()` controls the aesthetic style of the plots.
- **Parameters**:
  - `'darkgrid'`: Dark background with grid lines (default).
  - `'whitegrid'`: White background with grid lines.
  - `'dark'`: Dark background without grid lines.
  - `'white'`: White background without grid lines.
  - `'ticks'`: White background with axis ticks.

Here, we switch from `'ticks'` to `'darkgrid'`.

```python
sns.set_style('ticks')
sns.scatterplot(data=penguins, x='flipper_length_mm', y='body_mass_g', hue='island')
sns.despine()
```

### Removing Spines with `sns.despine()`

`sns.despine()` removes the top and right spines from the plot, which can give a cleaner look. It's often used with `'ticks'` or `'white'` styles.

```python
sns.set_style('ticks')
sns.set_context('notebook') #poster #paper #notebook #talk
sns.scatterplot(data=penguins, x='flipper_length_mm', y='body_mass_g', hue='island')
sns.despine()
```

### Setting Plot Context with `sns.set_context()`

`sns.set_context()` scales plot elements to better fit different presentation contexts (e.g., paper, notebook, talk, poster).
- **Parameters**:
  - `'paper'`: Smallest, for academic papers.
  - `'notebook'`: Default, for Jupyter notebooks.
  - `'talk'`: Larger, for presentations.
  - `'poster'`: Largest, for posters.

```python
sns.set_style('ticks')
sns.set_context('notebook') #poster #paper #notebook #talk
sns.scatterplot(data=penguins, x='flipper_length_mm', y='body_mass_g', hue='island', palette="Set1") #palette=set1,set2,set3
sns.despine()
```

### Customizing Color Palette

- `palette`: Specifies the color scheme for categorical mappings. Seaborn has many built-in palettes (e.g., `'deep'`, `'muted'`, `'bright'`, `'pastel'`, `'dark'`, `'colorblind'`, `'Set1'`, `'Set2'`, `'Set3'`, `'Paired'`, `'Accent'`, `'Dark2'`, `'tab10'`, `'tab20'`, etc.). You can also pass a list of colors.

```python
sns.scatterplot(data=penguins, x='species', y='body_mass_g', hue='island', style='sex', alpha=0.5)
```

### Scatter Plot with `style` and `alpha`

- `style`: Column name to vary marker styles based on categories. Adds another layer of categorical differentiation.
- `alpha`: Transparency of points (0 to 1). Useful for dense plots to show overlap.

```python
sns.stripplot(data=penguins, x='species', y='body_mass_g', hue='island', jitter=True)
```

## 5. Strip Plots

Strip plots show the distribution of a numerical variable across different categories by plotting individual observations as points. They are useful for small datasets or when combined with other plots.

```python
sns.stripplot(data=penguins, x='species', y='body_mass_g', hue='island')
```

### Basic Stripplot with `hue`

- `x`: Categorical variable.
- `y`: Numerical variable.
- `hue`: Categorical variable for coloring points.

Without `jitter=True`, points might overlap heavily, obscuring the distribution.

```python
sns.stripplot(data=penguins, x='species', y='body_mass_g', hue='island', dodge=True)
```

### Stripplot with `dodge`

- `dodge`: When `hue` is used, `True` separates the strip plots for different hue levels along the categorical axis, preventing overlap.

```python
sns.stripplot(data=penguins, x='species', y='body_mass_g', hue='island', dodge=True, jitter=True)
```

### Stripplot with `dodge` and `jitter`

- `jitter`: Adds random noise to the positions of the points along the categorical axis, which helps to spread out overlapping points and better visualize the density. Default is `False`.

```python
sns.swarmplot(data=penguins, x='species', y='body_mass_g', hue='island')
```

## 6. Swarm Plots

Swarm plots are similar to strip plots but arrange points to prevent overlap, providing a better representation of the distribution density. They are often more effective for larger datasets than strip plots.

```python
sns.set_context('notebook')
sns.histplot(data=penguins, x='body_mass_g',hue='sex')
```

## 7. Histograms (Histplots)

Histograms display the distribution of a single numerical variable by dividing the data into bins and counting the number of observations in each bin.

```python
sns.set_context('notebook')
sns.histplot(data=penguins, x='body_mass_g',hue='sex', multiple='stack')
```

### Histplot with `multiple='stack'`

- `multiple`: How to handle multiple distributions (when `hue` is used).
  - `'layer'`: (Default) Plot distributions on top of each other.
  - `'stack'`: Stack bars for each bin, showing total counts and proportions.
  - `'dodge'`: Place bars for each hue level side-by-side.
  - `'fill'`: Stack bars, normalized to 100%.

```python
sns.regplot(data=penguins, x='flipper_length_mm', y='body_mass_g',color='green') #scatter parameter to false for only line trend visual no data points
```

## 8. Regression Plots (Regplots)

Regression plots display the relationship between two numerical variables and fit a linear regression model to the data, along with a confidence interval around the regression line.

```python
sns.set_style('whitegrid')
sns.lineplot(data=penguins, x='body_mass_g', y='flipper_length_mm', hue='species', style='sex')
```

## 9. Line Plots

Line plots are typically used to show trends over a continuous variable (often time), but can also show relationships between two numerical variables, often with an aggregation (e.g., mean) and confidence intervals.

```python
sns.jointplot(data=penguins, y='flipper_length_mm', x='body_mass_g')
```

## 10. Joint Plots

Joint plots combine multiple plots to show the relationship between two variables and their individual distributions. They are very useful for exploring bivariate relationships.

```python
sns.jointplot(data=penguins, y='flipper_length_mm', x='body_mass_g', hue='sex', kind='kde')
```

### Joint Plot with `kind='kde'`

- `kind`: Type of plot to draw in the center and on the margins.
  - `'scatter'`: (Default) Scatter plot with histograms.
  - `'reg'`: Regression plot with histograms.
  - `'resid'`: Residuals plot with histograms.
  - `'kde'`: Kernel Density Estimate plot (contours) with KDE curves on margins.
  - `'hex'`: Hexbin plot with histograms.

```python
sns.jointplot(data=penguins, y='flipper_length_mm', x='body_mass_g', kind='reg')
```

### Joint Plot with `kind='reg'`

This shows a regression line in the center and histograms on the margins.

```python
sns.barplot(data=penguins, x='species', y='body_mass_g', hue='sex', palette=['skyblue', 'pink'], estimator=np.sum) #np.sum #np.mean #np.median #np.std #np.var #len #np.max #np.min
```

## 11. Bar Plots

Bar plots represent an estimate of central tendency (e.g., mean) for a numerical variable across different categories, and indicate the uncertainty around that estimate using error bars.

```python
sns.countplot(data=penguins, x='species', hue='island',palette="Set2")
```

## 12. Count Plots

Count plots show the counts of observations in each category of a categorical variable. They are essentially bar plots where the height of the bar shows the number of occurrences.

```python
sns.boxplot(data=penguins, x='species', y='body_mass_g', hue='island')
```

## 13. Box Plots

Box plots (or box-and-whisker plots) display the distribution of a numerical variable for one or more categories, showing quartiles (median, Q1, Q3) and potential outliers.

```python
sns.violinplot(data=penguins, x='species', y='body_mass_g', hue='island', split=True)
```

## 14. Violin Plots

Violin plots combine box plots with kernel density estimation (KDE) plots, showing both the summary statistics (like a box plot) and the distribution shape (like a KDE).

```python
sns.violinplot(data = penguins, x = "species", y ="body_mass_g",hue="species", palette = "pastel")
sns. swarmplot(data = penguins, x="species", y="body_mass_g", hue = "island", size=3, palette="dark")
```

### Combined Violin and Swarm Plot

This demonstrates how to layer different plot types on the same axes to provide richer information. Here, violin plots show the overall distribution, and swarm plots add individual data points on top.

Note: `sns.swarmplot` was manually corrected from `sns. swarmplot` in the original cell.

```python
sns.kdeplot(data=penguins, x='body_mass_g',  hue='species', fill=True)
```

## 15. Kernel Density Estimate (KDE) Plots

KDE plots visualize the probability density of a continuous variable. They are smoothed histograms, useful for showing the shape of a distribution.

```python
columns=["bill_length_mm","bill_depth_mm","flipper_length_mm",	"body_mass_g"]
penguins[columns].corr()
```

## 16. Correlation and Heatmaps

Correlation matrices show the pairwise correlation coefficients between numerical variables. Heatmaps are a visual representation of these matrices.

```python
sns.heatmap(data=penguins[columns].corr(), annot=True, cmap='Blues', vmin=-1, vmax=1)
```

### Correlation Heatmap

`sns.heatmap()`:
- `data`: The correlation matrix.
- `annot=True`: Displays the correlation values on the heatmap cells.
- `cmap='Blues'`: Color map to use for the heatmap (e.g., 'viridis', 'plasma', 'magma', 'cividis', 'Greens', 'Blues', 'Reds', etc.).
- `vmin`, `vmax`: Minimum and maximum values for the colormap, useful for consistent scaling (e.g., -1 to 1 for correlations).

```python
sns.rugplot(data=penguins, x='body_mass_g', y='flipper_length_mm', hue='species', height=0.2)
```

## 17. Rug Plots

Rug plots draw small marks along an axis for each observation, showing the distribution of a single numerical variable. They are often used to supplement other plots.

```python
sns.pairplot(data=penguins, hue='sex', diag_kind='hist')
```

## 18. Pair Plots

Pair plots create a grid of relationships between all pairs of numerical variables in a DataFrame, with distributions of individual variables on the diagonal. They are excellent for quickly exploring relationships in multivariate data.

```python
graph=sns.PairGrid(data=penguins, hue='sex')
graph.map_upper(sns.scatterplot)
graph.map_lower(sns.kdeplot)
graph.map_diag(sns.histplot)
graph.add_legend()
```

## 19. PairGrid

`sns.PairGrid` provides a more flexible way to create pair plots, allowing you to map different plotting functions to the upper, lower, and diagonal subplots. This gives fine-grained control over the visualization.
